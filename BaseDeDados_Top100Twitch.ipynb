# **Os 100 canais com mais seguidores da twitch**
Link: https://www.kaggle.com/datasets/mubashirsidiki/top-50-twitch-channels-by-followers


*   Quantidade de dados: 100;
*   Quantidade de variaveis: 3;
*   Rank: é a lista de todos os canais de 1 a 100;
*   Channel: É o nome do canal - Ex: KaiCenat;
*   Followers: É a quantidade de seguidores daquele canal - Ex:20.2M ;
*   Tipo de variaveis: qualitativa - KaiCenat. Quantitativa - 20.2M e 19.84M;

```python
df.sort_values(by='seguidores', ascending=True).head(15)
```

```python
df['seguidores'] = (
    df['seguidores']
    .str.replace('M', '', regex=False)
    .astype(float)
)
```

```python
df_top15 = (
    df.sort_values('seguidores', ascending=True)
      .head(15)
)
```

```python
plt.figure(figsize = (10, 5))
plt.barh(df_top15['canal'], df_top15['seguidores'], color = ['red', 'black'])
plt.title('Canais da Twitch com mais seguidores')
plt.ylabel('canal')
plt.xlabel('seguidores (Milhões)')
plt.xticks(rotation=45)
# plt.legend()
plt.show()
```
